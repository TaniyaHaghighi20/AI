```python
pExact=0.8
pOverShoot=0.1
pUnderShoot=0.1
# stochastic move
def Smove(p,U):
    q=[]
    for i in range(len(p)):
        s=pExact*p[(i-U)%len(p)]
        s+=pOverShoot*p[(i-U-1)%len(p)]
        s+=pUnderShoot*p[(i-U+1)%len(p)]
        q.append(s)
    return q
p=[0,0.5,0,0.5,0]
Smove(p,2)
```
